```python
import pandas as pd
import sqlite3

#Aqui agrega la conexión a la base de datos
conn=sqlite3.connect("Salaries.sqlite")

#Crea el dataframe con los datos del datasets
df=pd.read_sql_query("SELECT * FROM salaries",conn)
# print(df.head(5))
```

1. **Top 10 empleados con mayor salario total (incluyendo beneficios)**
Extrae e imprime un top 10 de expleados con mayor salario incluyendo los beneficios 


```python
# Actividad 1:
# convertimos las columnas de objetos a float
df["TotalPayBenefits"] = pd.to_numeric(df["TotalPayBenefits"], errors="coerce")
df["OvertimePay"] = pd.to_numeric(df["OvertimePay"], errors="coerce")
df["BasePay"] = pd.to_numeric(df["BasePay"], errors="coerce")
df["TotalPay"] = pd.to_numeric(df["TotalPay"], errors="coerce")

# hacemos la sentencia sql para el filtrado y traemos los 10 más pagados
df_salaries=pd.read_sql_query("SELECT EmployeeName, TotalPay, TotalPayBenefits FROM salaries", conn)
top_10 = df_salaries.sort_values(by='TotalPayBenefits', ascending=False).head(10)
print(f"los 10 empleados con mayor salario son: \n {top_10}")
# conn.close()
```

2. **Filtrar empleados con más de 50,000 en horas extra, la columna a tratar es: (OvertimePay)**

```python
# Actividad 2:
# traemos los empleados con más de horas extras trabajadas y mostramos los primeros 5
df_overtime=df[df['OvertimePay']>50000].head()
print(f"los empleados con más de 50mil horas extra son: \n {df_overtime}")
```

3. **Contar cuántos empleados únicos hay por año**

```python
# Actividad 3:
# se agrupan los datos por año y nombre de empleado
empleados_unicos = df.groupby('Year')['EmployeeName'].nunique()
print(f"la cantidad de empleados únicos que hay por año son: \n {empleados_unicos}")
```

4. **Ver cuántos cargos únicos (JobTitle) existen y los 5 más comunes**

```python
# Actividad 4:
cargos_unicos=df["JobTitle"].nunique()
print(f"los cantidad de cargos únicos que existen son: {cargos_unicos}")

mas_comunes=df["JobTitle"].value_counts().head(6)
print(f"\n los trabajos más comúnes son: {mas_comunes}")
```

5. **Mostrar el salario total (TotalPay) promedio por año**

```python
# Actividad 5:

# se agrupa por año y el salario total y se saca los beneficios utilizando el método mean()
salario_promedio=df.groupby("Year")["TotalPay"].mean()
print(f"el promedio de salario total por año es: {salario_promedio}")
```

6. **Exporta todos los datos a un CSV y muestra los ultimos 10 valores**

```python
# Actividad 6:
#Exportar el DataFrame a un archivo CSV
df.to_csv("new_salaries.csv", index=False)
print(f"Últimos 10 registros del DataFrame exportado: \n {df.tail(10)}") 
```
